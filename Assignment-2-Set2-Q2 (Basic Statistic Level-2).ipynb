# Basic Statistic Level-2 Set-2 Q-2

```python
from scipy import stats
from scipy.stats import norm
```

```python
# A. More employees at the processing center are older than 44 than between 38 and 44.
```

```python
# p(X>44) : Employees older than 44 years of age
1-stats.norm.cdf(44,loc=38,scale=6)  # loc = mean , scale = standard deviation
```

```python
# p(38<X<44) : Employees between 38 to 44 years of age
stats.norm.cdf(44,loc=38,scale=6)-stats.norm.cdf(38,loc=38,scale=6)
```

```python
# B. A training program for employees under the age of 30 at the center would be expected to attract about 36 employees.
```

```python
# P(X<30) : Employees under 30 years of age
stats.norm.cdf(30,loc=38,scale=6)
```

```python
# No. of employees attending training program from 400 no's is N*P(X<30)
400*stats.norm.cdf(30,loc=38,scale=6)
```
